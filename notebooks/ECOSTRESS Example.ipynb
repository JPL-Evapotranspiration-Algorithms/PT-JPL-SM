```python
results = PTJPLSM(
    geometry=geometry,
    time_UTC=time_UTC,
    NDVI=NDVI,
    Ta_C=Ta_C,
    RH=RH,
    Rn_Wm2=Rn,
    ST_C=ST_C,
    albedo=albedo,
    upscale_to_daylight=True
)
```

```python
LE_Wm2 = results["LE_Wm2"]
LE_Wm2.cmap = ET_COLORMAP
LE_Wm2
```

```python
ET_daylight_kg = results["ET_daylight_kg"]
ET_daylight_kg.cmap = ET_COLORMAP
ET_daylight_kg
```

```python
ET_to_LE_scale_factor = LE_Wm2 / ET_daylight_kg
ET_to_LE_scale_factor
```

```python
ET_to_LE_scale_factor_mean = np.nanmean(ET_to_LE_scale_factor)
ET_to_LE_scale_factor_mean
```

```python
LE_scaled_by_mean = ET_daylight_kg * ET_to_LE_scale_factor_mean
LE_scaled_by_mean
```

```python
scaling_error = (LE_scaled_by_mean - LE_Wm2) / LE_Wm2 * 100
scaling_error
```
